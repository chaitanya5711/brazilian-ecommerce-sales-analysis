```python
#Step 5.1 — Create the main analysis table
```

```python
orders_clean = orders.copy()

print("Orders clean created successfully!")
print("Shape:", orders_clean.shape)
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders_clean[col] = pd.to_datetime(
        orders_clean[col],
        errors="coerce"
    )

print("Date columns converted successfully!")
```

```python
order_items_clean = order_items.copy()

order_items_clean["total_item_value"] = (
    order_items_clean["price"] +
    order_items_clean["freight_value"]
)

print("Order Items clean created successfully!")
print("Shape:", order_items_clean.shape)
```

```python
sales_data = order_items_clean.merge(
    orders_clean,
    on="order_id",
    how="left"
)

print("After merging Orders:", sales_data.shape)
```

```python
#Step 5.2 — Merge Customers
```

```python
sales_data = sales_data.merge(
    customers_clean,
    on="customer_id",
    how="left"
)

print("After merging Customers:", sales_data.shape)
```

```python
sales_data.head()
```

```python
#Step 5.3 — Merge Products
```

```python
sales_data = sales_data.merge(
    products,
    on="product_id",
    how="left"
)

print("After merging Products:", sales_data.shape)
```

```python
sales_data.columns
```

```python
#Step 5.4 — Merge Sellers
```

```python
sales_data = sales_data.merge(
    sellers,
    on="seller_id",
    how="left"
)

print("Final sales_data shape:", sales_data.shape)
```

```python
sales_data.head()
```

```python
#Step 6.1 — Basic Business KPIs
```

```python
#Cell 1 — Total Orders
total_orders = sales_data["order_id"].nunique()

print("Total Orders:", total_orders)
```

```python
#Cell 2 — Total Customers
total_customers = sales_data["customer_unique_id"].nunique()

print("Total Customers:", total_customers)
```

```python
#Cell 3 — Total Products Sold
total_products_sold = sales_data["order_item_id"].count()

print("Total Products Sold:", total_products_sold)
```

```python
#Cell 4 — Total Revenue
total_revenue = sales_data["price"].sum()

print("Total Revenue: ₹", round(total_revenue, 2))
```

```python
#Cell 5 — Total Freight
total_freight = sales_data["freight_value"].sum()

print("Total Freight (BRL):", round(total_freight, 2))
```

```python
#Cell 6 — Average Order Value
average_order_value = (
    sales_data.groupby("order_id")["price"]
    .sum()
    .mean()
)

print("Average Order Value (BRL):", round(average_order_value, 2))
```

```python
#Cell 7 — Put all KPIs together
print("========== OLIST E-COMMERCE KPIs ==========")
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Products Sold:", total_products_sold)
print("Total Revenue (BRL):", round(total_revenue, 2))
print("Total Freight (BRL):", round(total_freight, 2))
print("Average Order Value (BRL):", round(average_order_value, 2))
```

```python
#Step 6.7 — Delivery Performance 
```

```python
orders_clean["delivery_days"] = (
    orders_clean["order_delivered_customer_date"]
    - orders_clean["order_purchase_timestamp"]
).dt.days

print("delivery_days created successfully!")
```

```python
#Cell 2 —  delivery delay
orders_clean["delivery_delay_days"] = (
    orders_clean["order_delivered_customer_date"]
    - orders_clean["order_estimated_delivery_date"]
).dt.days

print("delivery_delay_days created successfully!")
```

```python
#Cell 3 —  delivery status
orders_clean["delivery_status"] = np.where(
    orders_clean["order_delivered_customer_date"].isna(),
    "Not Delivered",
    np.where(
        orders_clean["delivery_delay_days"] > 0,
        "Late",
        "On Time"
    )
)

print(orders_clean["delivery_status"].value_counts())
```

```python
sales_data = sales_data.drop(
    columns=["delivery_days", "delivery_delay_days", "delivery_status"],
    errors="ignore"
)

sales_data = sales_data.merge(
    orders_clean[
        [
            "order_id",
            "delivery_days",
            "delivery_delay_days",
            "delivery_status"
        ]
    ],
    on="order_id",
    how="left"
)

print("Delivery columns added successfully!")
```

```python
print(sales_data[
    [
        "order_id",
        "delivery_days",
        "delivery_delay_days",
        "delivery_status"
    ]
].head())
```

```python
#Cell 6 — Average delivery time
average_delivery_days = sales_data["delivery_days"].mean()

print(
    "Average Delivery Time:",
    round(average_delivery_days, 2),
    "days"
)
```

```python
#Step 6.8 — Customer Analysis.
```

```python
#Cell 1 — Orders per customer
customer_orders = (
    sales_data.groupby("customer_unique_id")["order_id"]
    .nunique()
)

print(customer_orders.describe())
```

```python
#Cell 2 — New vs Repeat Customers
new_customers = (customer_orders == 1).sum()
repeat_customers = (customer_orders > 1).sum()

print("One-time Customers:", new_customers)
print("Repeat Customers:", repeat_customers)
```

```python
#Cell 3 — Repeat customer percentage
repeat_customer_percentage = (
    repeat_customers / len(customer_orders) * 100
)

print(
    "Repeat Customer Percentage:",
    round(repeat_customer_percentage, 2),
    "%"
)
```

```python
#Cell 4 — Customer purchase frequency
purchase_frequency = (
    customer_orders
    .value_counts()
    .sort_index()
)

print(purchase_frequency)
```

```python
#Cell 5 — Visualization
plt.figure(figsize=(10, 6))

purchase_frequency.head(10).plot(kind="bar")

plt.title("Customer Purchase Frequency")
plt.xlabel("Number of Orders")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
